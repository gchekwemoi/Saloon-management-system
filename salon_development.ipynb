```python

from decimal import Decimal

# Create two employees
employee1 = Employee("E001", "Employee A", "Braiding")
employee2 = Employee("E002", "Employee B", "Braiding")

# One client's agreed braiding price
price = Decimal("150000")

# Assign employee contributions and wage rates
assignment1 = ServiceAssignment(
    employee1, Decimal("60"), Decimal("30")
)

assignment2 = ServiceAssignment(
    employee2, Decimal("40"), Decimal("35")
)

# Calculate revenue contributions and expected wages
for assignment in [assignment1, assignment2]:
    revenue = assignment.calculate_contribution(price)
    wage = assignment.calculate_expected_wage(price)

    print(f"\nEmployee: {assignment.employee.name}")
    print(f"Revenue contribution: UGX {revenue:,.0f}")
    print(f"Wage rate: {assignment.wage_rate}%")
    print(f"Expected wage: UGX {wage:,.0f}")
```

```python

from datetime import date
from decimal import Decimal
from src.models.product_sale import ProductSale

# Record three different product sales
products = [
    ProductSale(
        sale_id="P001",
        sale_date=date.today(),
        product_description="Black braids, long pack",
        quantity=3,
        unit_selling_price=Decimal("8000")
    ),
    ProductSale(
        sale_id="P002",
        sale_date=date.today(),
        product_description="Assorted beads",
        quantity=2,
        unit_selling_price=Decimal("2500")
    ),
    ProductSale(
        sale_id="P003",
        sale_date=date.today(),
        product_description="Beauty hair oil",
        quantity=2,
        unit_selling_price=Decimal("15000")
    )
]

# Display product sales
total_revenue = Decimal("0")

for product in products:
    amount = product.calculate_total()
    total_revenue += amount

    print(f"{product.product_description}: UGX {amount:,.0f}")

print("-" * 35)
print(f"TOTAL PRODUCT REVENUE: UGX {total_revenue:,.0f}")
```
